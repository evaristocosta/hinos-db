Problemas encontrados podem ser corrigidos aqui, adicionados ao arquivo de migração `009-fix-hinos.sql`.

```python
from sqlalchemy import create_engine

eng_hdb = create_engine("sqlite:///../../database/database.db")
con_hdb = eng_hdb.connect()
```

```python
import pandas as pd

sql_fix = """
select cast(hc.hino_numero as integer) as numero, coletanea_id, count(*) as total 
  from hino h
  inner join hino_coletanea hc on hc.hino_id = h.id
  group by hc.hino_numero, coletanea_id
  having total > 1;
"""
hino_fix = pd.read_sql_query(sql_fix, con_hdb)
hino_fix.sort_values("numero", ascending=True)
```

```python
con_hdb.close()
eng_hdb.dispose()
```
